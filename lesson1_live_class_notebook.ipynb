````python
#@title 📦 Data Types (click ▶ to re-render if needed)
from IPython.display import HTML, display
display(HTML("""
<div id="dt_widget" style="
  max-width:620px; margin:16px auto; font-family:'Segoe UI',system-ui,sans-serif;
  background:linear-gradient(135deg,#1a1a2e 0%,#16213e 50%,#0f3460 100%);
  border-radius:14px; border:1px solid rgba(99,179,237,0.15);
  overflow:hidden; user-select:none;
">
  <!-- Title bar -->
  <div style="padding:14px 20px 6px; display:flex; justify-content:space-between; align-items:center;">
    <span style="color:#90cdf4; font-size:13px; letter-spacing:1.5px; text-transform:uppercase; font-weight:600;">📦 Data Types</span>
    <span id="dt_widget_counter" style="color:#4a5568; font-size:12px; font-family:monospace;"></span>
  </div>

  <!-- Visual area -->
  <div id="dt_widget_stage" style="padding:0 10px; min-height:340px; display:flex; align-items:center; justify-content:center;">
  </div>

  <!-- Controls -->
  <div style="padding:10px 20px 14px; display:flex; justify-content:center; align-items:center; gap:12px;">
    <button id="dt_widget_prev" onclick="dt_widget_nav(-1)" style="
      background:rgba(99,179,237,0.1); border:1px solid rgba(99,179,237,0.3); color:#90cdf4;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.25)'" onmouseout="this.style.background='rgba(99,179,237,0.1)'">
      ◀ Prev
    </button>

    <div id="dt_widget_dots" style="display:flex; gap:6px; align-items:center;"></div>

    <button id="dt_widget_next" onclick="dt_widget_nav(1)" style="
      background:rgba(99,179,237,0.2); border:1px solid rgba(99,179,237,0.4); color:#fff;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.4)'" onmouseout="this.style.background='rgba(99,179,237,0.2)'">
      Next ▶
    </button>
  </div>
</div>

<script>
(function() {
  const W = 600, H = 340;
  const id = "dt_widget";
  let step = 0;
  const steps = [`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="94" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="231" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="368" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="505" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(99,179,237,0.12)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="94" y="112" fill="rgb(99,179,237)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">int</text><text x="94" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Whole Numbers</text><line x1="42" y1="148" x2="147" y2="148" stroke="#4a5568" stroke-width="1"/><text x="42" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">num_stops = 12</text><text x="42" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">passengers = 48</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="231" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="368" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="505" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><text x="300" y="260" fill="rgb(99,179,237)" font-size="15" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">▲ int — Whole Numbers</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(99,179,237,0.12)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="94" y="112" fill="rgb(99,179,237)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">int</text><text x="94" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Whole Numbers</text><line x1="42" y1="148" x2="147" y2="148" stroke="#4a5568" stroke-width="1"/><text x="42" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">num_stops = 12</text><text x="42" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">passengers = 48</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(154,230,180,0.12)" stroke="rgb(154,230,180)" stroke-width="2" opacity="1"/><text x="231" y="112" fill="rgb(154,230,180)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">float</text><text x="231" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Decimal Numbers</text><line x1="179" y1="148" x2="284" y2="148" stroke="#4a5568" stroke-width="1"/><text x="179" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">base_fare = 18.50</text><text x="179" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">distance  = 7.3</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="368" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="505" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><text x="300" y="260" fill="rgb(154,230,180)" font-size="15" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">▲ float — Decimal Numbers</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(99,179,237,0.12)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="94" y="112" fill="rgb(99,179,237)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">int</text><text x="94" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Whole Numbers</text><line x1="42" y1="148" x2="147" y2="148" stroke="#4a5568" stroke-width="1"/><text x="42" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">num_stops = 12</text><text x="42" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">passengers = 48</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(154,230,180,0.12)" stroke="rgb(154,230,180)" stroke-width="2" opacity="1"/><text x="231" y="112" fill="rgb(154,230,180)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">float</text><text x="231" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Decimal Numbers</text><line x1="179" y1="148" x2="284" y2="148" stroke="#4a5568" stroke-width="1"/><text x="179" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">base_fare = 18.50</text><text x="179" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">distance  = 7.3</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(183,148,244,0.12)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="368" y="112" fill="rgb(183,148,244)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">str</text><text x="368" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Text / Strings</text><line x1="316" y1="148" x2="421" y2="148" stroke="#4a5568" stroke-width="1"/><text x="316" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">route = &quot;12A&quot;</text><text x="316" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">stop  = &quot;Central&quot;</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(74,85,104,0.08)" stroke="#2d3748" stroke-width="2" opacity="1"/><text x="505" y="165" fill="#4a5568" font-size="24" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">?</text><text x="300" y="260" fill="rgb(183,148,244)" font-size="15" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">▲ str — Text / Strings</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(99,179,237,0.12)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="94" y="112" fill="rgb(99,179,237)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">int</text><text x="94" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Whole Numbers</text><line x1="42" y1="148" x2="147" y2="148" stroke="#4a5568" stroke-width="1"/><text x="42" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">num_stops = 12</text><text x="42" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">passengers = 48</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(154,230,180,0.12)" stroke="rgb(154,230,180)" stroke-width="2" opacity="1"/><text x="231" y="112" fill="rgb(154,230,180)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">float</text><text x="231" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Decimal Numbers</text><line x1="179" y1="148" x2="284" y2="148" stroke="#4a5568" stroke-width="1"/><text x="179" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">base_fare = 18.50</text><text x="179" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">distance  = 7.3</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(183,148,244,0.12)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="368" y="112" fill="rgb(183,148,244)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">str</text><text x="368" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Text / Strings</text><line x1="316" y1="148" x2="421" y2="148" stroke="#4a5568" stroke-width="1"/><text x="316" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">route = &quot;12A&quot;</text><text x="316" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">stop  = &quot;Central&quot;</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(251,211,141,0.12)" stroke="rgb(251,211,141)" stroke-width="2" opacity="1"/><text x="505" y="112" fill="rgb(251,211,141)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">bool</text><text x="505" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">True / False</text><line x1="453" y1="148" x2="558" y2="148" stroke="#4a5568" stroke-width="1"/><text x="453" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_peak = True</text><text x="453" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_ac   = False</text><text x="300" y="260" fill="rgb(251,211,141)" font-size="15" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">▲ bool — True / False</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="30" fill="#63b3ed" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">CITYSMART DATA DEPOT</text><text x="300" y="55" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every value in Python has a TYPE</text><rect x="32" y="90" width="125" height="145" rx="10" fill="rgba(99,179,237,0.12)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="94" y="112" fill="rgb(99,179,237)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">int</text><text x="94" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Whole Numbers</text><line x1="42" y1="148" x2="147" y2="148" stroke="#4a5568" stroke-width="1"/><text x="42" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">num_stops = 12</text><text x="42" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">passengers = 48</text><rect x="169" y="90" width="125" height="145" rx="10" fill="rgba(154,230,180,0.12)" stroke="rgb(154,230,180)" stroke-width="2" opacity="1"/><text x="231" y="112" fill="rgb(154,230,180)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">float</text><text x="231" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Decimal Numbers</text><line x1="179" y1="148" x2="284" y2="148" stroke="#4a5568" stroke-width="1"/><text x="179" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">base_fare = 18.50</text><text x="179" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">distance  = 7.3</text><rect x="306" y="90" width="125" height="145" rx="10" fill="rgba(183,148,244,0.12)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="368" y="112" fill="rgb(183,148,244)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">str</text><text x="368" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Text / Strings</text><line x1="316" y1="148" x2="421" y2="148" stroke="#4a5568" stroke-width="1"/><text x="316" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">route = &quot;12A&quot;</text><text x="316" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">stop  = &quot;Central&quot;</text><rect x="443" y="90" width="125" height="145" rx="10" fill="rgba(251,211,141,0.12)" stroke="rgb(251,211,141)" stroke-width="2" opacity="1"/><text x="505" y="112" fill="rgb(251,211,141)" font-size="20" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">bool</text><text x="505" y="135" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">True / False</text><line x1="453" y1="148" x2="558" y2="148" stroke="#4a5568" stroke-width="1"/><text x="453" y="168" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_peak = True</text><text x="453" y="188" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_ac   = False</text><rect x="60" y="260" width="480" height="40" rx="8" fill="rgba(251,211,141,0.08)" stroke="rgba(251,211,141,0.3)" stroke-width="2" opacity="1"/><text x="300" y="282" fill="#fbd38d" font-size="14" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">The TYPE determines what OPERATIONS are allowed</text></svg>`];

  const stage = document.getElementById(id + "_stage");
  const counter = document.getElementById(id + "_counter");
  const dots_el = document.getElementById(id + "_dots");
  const prev_btn = document.getElementById(id + "_prev");
  const next_btn = document.getElementById(id + "_next");

  // Build dots
  for (let i = 0; i < steps.length; i++) {
    const d = document.createElement("div");
    d.style.cssText = "width:8px;height:8px;border-radius:50%;background:#4a5568;transition:all 0.2s;cursor:pointer;";
    d.onclick = function() { step = i; render(); };
    dots_el.appendChild(d);
  }

  function render() {
    stage.innerHTML = steps[step];
    counter.textContent = "Step " + (step+1) + " of " + steps.length;
    prev_btn.style.opacity = step === 0 ? "0.3" : "1";
    prev_btn.style.pointerEvents = step === 0 ? "none" : "auto";
    next_btn.style.opacity = step === steps.length-1 ? "0.3" : "1";
    next_btn.style.pointerEvents = step === steps.length-1 ? "none" : "auto";
    // Update dots
    for (let i = 0; i < dots_el.children.length; i++) {
      dots_el.children[i].style.background = i === step ? "#63b3ed" : "#4a5568";
      dots_el.children[i].style.transform = i === step ? "scale(1.4)" : "scale(1)";
    }
  }

  window[id + "_nav"] = function(dir) {
    step = Math.max(0, Math.min(steps.length-1, step + dir));
    render();
  };

  // Keyboard support
  document.addEventListener("keydown", function(e) {
    const rect = document.getElementById(id).getBoundingClientRect();
    const inView = rect.top < window.innerHeight && rect.bottom > 0;
    if (!inView) return;
    if (e.key === "ArrowRight" || e.key === " ") { e.preventDefault(); window[id + "_nav"](1); }
    if (e.key === "ArrowLeft") { e.preventDefault(); window[id + "_nav"](-1); }
  });

  render();
})();
</script>
"""))
````

### DEMO — The 4 core data types in CitySmart
### Run this cell and observe the output carefully
#### ============================================================

## --- int: whole numbers, used for counts and IDs ---


````python
route_number   = 12          # bus route number
num_stops      = 18          # total stops on the route
passengers     = 43          # passengers on current trip
````

##--- float: decimal numbers, used for money and distances ---


````python
base_fare      = 18.50       # base fare in rupees
distance_km    = 11.3        # route length
fuel_cost      = 4.75        # cost per km
````

## --- str: text, used for names, codes, messages ---


````python
route_name     = "12A Express"   # route display name
start_stop     = "Central Bus Stand"
end_stop       = "Airport Terminal 2"
````

## --- bool: True or False only, used for flags and conditions ---


````python
is_peak_hour   = True        # is this a peak-hour trip?
is_ac_bus      = False       # does this bus have AC?
is_accessible  = True        # wheelchair accessible?
````

## --- type() tells you what type a variable is ---


````python
print("=== CitySmart Route Data ===")
print(f"Route:      {route_name}  →  type: {type(route_name)}")
print(f"Stops:      {num_stops}              →  type: {type(num_stops)}")
print(f"Base fare:  {base_fare}           →  type: {type(base_fare)}")
print(f"Peak hour:  {is_peak_hour}          →  type: {type(is_peak_hour)}")
````

## --- Dynamic typing in action ---


````python
print("\n=== Dynamic Typing Demo ===")
fare = 18       # starts as int
print(f"fare = {fare}    type: {type(fare)}")
fare = 18.50    # now it's a float
print(f"fare = {fare}  type: {type(fare)}")
fare = "free"   # now it's a string (e.g. for children under 5)
print(f"fare = {fare}   type: {type(fare)}")
````

### 🔀 Type Casting — converting between types

Sometimes Python gives you data in the wrong type — especially when reading user input (which is always a string) or CSV files. **Type casting** converts one type into another.

```
  TYPE CASTING IN CITYSMART
  ─────────────────────────────────────────────────────────────

  User types "42" at the terminal
         │
         ▼
  input() always returns str ──→ "42"  (this is TEXT, not a number!)
         │
         │  int("42")  ──────────→   42  ✓ now you can do maths
         │  float("42") ─────────→ 42.0  ✓ decimal version
         │  str(42) ────────────→ "42"   ✓ number → text for display
         │  bool(0) ────────────→ False  ✓ zero = False, rest = True
         │
  DANGER ZONE:
         │  int("twelve")  ──────→ ❌ ValueError: cannot convert
         │  int("18.50")   ──────→ ❌ ValueError: use float() first
         │  float("18.50") ──────→ 18.5  ✓ then int(18.5) → 18
  ─────────────────────────────────────────────────────────────
```

## Simulating data that arrived as strings (e.g. from a CSV or user input)


````python
raw_passengers = "43"       # str — came from a CSV file
raw_fare       = "18.50"    # str — same
raw_is_peak    = "1"        # str — 1 means True in the dataset
````

## Convert to correct types before doing any calculations


````python
passengers  = int(raw_passengers)       # "43" → 43
fare        = float(raw_fare)           # "18.50" → 18.5
is_peak     = bool(int(raw_is_peak))    # "1" → 1 → True
````

````python
print("--- After type casting ---")
print(f"Passengers : {passengers}  ({type(passengers).__name__})")
print(f"Fare       : {fare}  ({type(fare).__name__})")
print(f"Is peak?   : {is_peak}  ({type(is_peak).__name__})")
````

## Now we can do maths — this would FAIL if we hadn't cast


````python
total_revenue = passengers * fare
print(f"\nTotal revenue from this trip: ₹{total_revenue}")
````

## Raw data from the bus sensor (everything is a string)

````python
raw_route_id      = "7"          # should become: ???
raw_route_name    = "Harbour Express"  # should become: ???
raw_passengers    = "52"         # should become: ???
raw_base_fare     = "22.50"      # should become: ???
raw_stops_taken   = "9"          # how many stops the passenger travelled
raw_is_student    = "1"          # 1 = student, 0 = regular passenger
raw_is_ac         = "0"          # 1 = AC bus, 0 = non-AC
````

### --- TASK 1: Cast each variable to the correct type ---


````python
# Replace ??? with the correct type conversion function

route_id     = int(raw_route_id)     # whole number
route_name   = raw_route_name        # already a string, no cast needed
passengers   = int(raw_passengers)   # whole number
base_fare    = float(raw_base_fare)     # decimal number
stops_taken  = int(raw_stops_taken)  # whole number
is_student   = bool(int( raw_is_student))  # True/False (hint: two-step cast)
is_ac        = bool(int(raw_is_ac))        # True/False
````

### --- TASK 2: Calculate the total fare ---
### Rules:
#####   - Base fare applies for the first 5 stops
#####   - Each additional stop costs ₹2.50 extra
#####   - Students get a 50% discount on the total
#####**bold text**  - AC buses add ₹5 flat surcharge (applied AFTER student discount)

````python
print(is_student)
````

````python
print(is_student * 0.5)
````

````python


extra_stops  = max(stops_taken - 5, 0)        # stops beyond the first 5 (min 0)
#we don't want a -ve value in extra_stops, using MAX func()

extra_cost   = extra_stops * 2.50                   # extra_stops × 2.50
total_before_discount = base_fare + extra_cost  # base_fare + extra_cost

discount     = is_student * 0.5                 # 50% if student else 0
#python will TYPECAST this value as int-> 1

fare_after_discount = total_before_discount * ( 1- discount ) # total_before_discount - discount
ac_surcharge = is_ac * 5                   # 5 if AC bus else 0
final_fare   = fare_after_discount + ac_surcharge       # fare_after_discount + ac_surcharge


````

### --- TASK 3: Print a receipt ---




````python
### Expected output should look like:

###   ============================
##    CITYSMART TRIP RECEIPT
###   ============================
###   Route     : 7 - Harbour Express
###   Passengers: 52
###   Stops     : 9
###  Student?  : Yes
###   AC Bus?   : No
###   ----------------------------
###   Base fare : ₹22.50
###   Extra     : ₹10.00
###   Discount  : -₹16.25
###   AC charge : ₹0.00
###   TOTAL     : ₹16.25
###   ============================

## Write your print statements here:
print ("="*28)
print("  CITYSMART TRIP RECEIPT")
print ("="*28)
print ( f" Route     : {route_id} - {route_name}")
print ( f" Passengers: {passengers}")
print ( f" Stops: {stops_taken}")
# print ( f" Student? : {'Yes' if is_student else 'No' }" )
print ( f" Student? : {'Yes' * (is_student) + 'No' * (1- is_student) }")
print ( f" AC Bus? : {'Yes' * is_ac + 'No' * (1- is_ac) }")
print ( "-" * 28)
print ( f" Base fare: ₹{base_fare: .2f} ")
print (f" Extra: ₹{extra_cost: .2f}")
print (f" Discount: -₹{total_before_discount - final_fare: .2f}")
print (f" AC charge: ₹{ac_surcharge: .2f}")
print (f" TOTAL: ₹{final_fare: .2f}")
print ("="*28)
````

````python
#@title ⚖️ Comparison Ops (click ▶ to re-render if needed)
from IPython.display import HTML, display
display(HTML("""
<div id="comp_widget" style="
  max-width:620px; margin:16px auto; font-family:'Segoe UI',system-ui,sans-serif;
  background:linear-gradient(135deg,#1a1a2e 0%,#16213e 50%,#0f3460 100%);
  border-radius:14px; border:1px solid rgba(99,179,237,0.15);
  overflow:hidden; user-select:none;
">
  <!-- Title bar -->
  <div style="padding:14px 20px 6px; display:flex; justify-content:space-between; align-items:center;">
    <span style="color:#90cdf4; font-size:13px; letter-spacing:1.5px; text-transform:uppercase; font-weight:600;">⚖️ Comparison Operators</span>
    <span id="comp_widget_counter" style="color:#4a5568; font-size:12px; font-family:monospace;"></span>
  </div>

  <!-- Visual area -->
  <div id="comp_widget_stage" style="padding:0 10px; min-height:340px; display:flex; align-items:center; justify-content:center;">
  </div>

  <!-- Controls -->
  <div style="padding:10px 20px 14px; display:flex; justify-content:center; align-items:center; gap:12px;">
    <button id="comp_widget_prev" onclick="comp_widget_nav(-1)" style="
      background:rgba(99,179,237,0.1); border:1px solid rgba(99,179,237,0.3); color:#90cdf4;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.25)'" onmouseout="this.style.background='rgba(99,179,237,0.1)'">
      ◀ Prev
    </button>

    <div id="comp_widget_dots" style="display:flex; gap:6px; align-items:center;"></div>

    <button id="comp_widget_next" onclick="comp_widget_nav(1)" style="
      background:rgba(99,179,237,0.2); border:1px solid rgba(99,179,237,0.4); color:#fff;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.4)'" onmouseout="this.style.background='rgba(99,179,237,0.2)'">
      Next ▶
    </button>
  </div>
</div>

<script>
(function() {
  const W = 600, H = 340;
  const id = "comp_widget";
  let step = 0;
  const steps = [`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="35" fill="#76e4f7" font-size="24" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="65" fill="#718096" font-size="14" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every comparison returns either True or False</text><rect x="20" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="55" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="150" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="245" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="340" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="435" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="110" width="75" height="55" rx="8" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="530" y="132" fill="#76e4f7" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="152" fill="#718096" font-size="10" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text><text x="300" y="210" fill="#718096" font-size="14" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Click Next to evaluate each one →</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">passengers &gt; 40</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">43 &gt; 40</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(154,230,180,0.15)" stroke="rgb(154,230,180)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(154,230,180)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">True</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Is the bus crowded?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">fare == 0</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">18.5 == 0</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(252,129,129,0.15)" stroke="rgb(252,129,129)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(252,129,129)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">False</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Is the trip free?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_id != 12</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">7 != 12</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(154,230,180,0.15)" stroke="rgb(154,230,180)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(154,230,180)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">True</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Is this NOT route 12?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">stops &gt;= 5</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">11 &gt;= 5</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(154,230,180,0.15)" stroke="rgb(154,230,180)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(154,230,180)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">True</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Qualifies for extra charge?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">revenue &lt; 500</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">795 &lt; 500</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(252,129,129,0.15)" stroke="rgb(252,129,129)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(252,129,129)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">False</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Below daily target?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#76e4f7" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Comparison Operators</text><text x="300" y="52" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Always return True or False</text><rect x="30" y="80" width="330" height="85" rx="10" fill="rgba(118,228,247,0.06)" stroke="rgba(118,228,247,0.2)" stroke-width="2" opacity="1"/><text x="50" y="108" fill="#e2e8f0" font-size="18" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">load &lt;= capacity</text><text x="50" y="138" fill="#718096" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">43 &lt;= 60</text><line x1="370" y1="122" x2="400" y2="122" stroke="#a0aec0" stroke-width="2"/><polygon points="410,122 400,117 400,127" fill="#a0aec0"/><rect x="420" y="88" width="150" height="68" rx="12" fill="rgba(154,230,180,0.15)" stroke="rgb(154,230,180)" stroke-width="3" opacity="1"/><text x="495" y="125" fill="rgb(154,230,180)" font-size="26" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">True</text><text x="300" y="205" fill="#a0aec0" font-size="15" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Within safe capacity?</text><rect x="20" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="55" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;</text><text x="55" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater</text><rect x="115" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="150" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">==</text><text x="150" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">equal</text><rect x="210" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="245" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">!=</text><text x="245" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">not equal</text><rect x="305" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="340" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&gt;=</text><text x="340" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">greater/eq</text><rect x="400" y="235" width="75" height="48" rx="6" fill="transparent" stroke="transparent" stroke-width="1" opacity="1"/><text x="435" y="255" fill="#4a5568" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;</text><text x="435" y="273" fill="#2d3748" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less</text><rect x="495" y="235" width="75" height="48" rx="6" fill="rgba(118,228,247,0.15)" stroke="rgba(118,228,247,0.4)" stroke-width="1" opacity="1"/><text x="530" y="255" fill="#76e4f7" font-size="16" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">&lt;=</text><text x="530" y="273" fill="#76e4f7" font-size="9" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">less/eq</text></svg>`];

  const stage = document.getElementById(id + "_stage");
  const counter = document.getElementById(id + "_counter");
  const dots_el = document.getElementById(id + "_dots");
  const prev_btn = document.getElementById(id + "_prev");
  const next_btn = document.getElementById(id + "_next");

  // Build dots
  for (let i = 0; i < steps.length; i++) {
    const d = document.createElement("div");
    d.style.cssText = "width:8px;height:8px;border-radius:50%;background:#4a5568;transition:all 0.2s;cursor:pointer;";
    d.onclick = function() { step = i; render(); };
    dots_el.appendChild(d);
  }

  function render() {
    stage.innerHTML = steps[step];
    counter.textContent = "Step " + (step+1) + " of " + steps.length;
    prev_btn.style.opacity = step === 0 ? "0.3" : "1";
    prev_btn.style.pointerEvents = step === 0 ? "none" : "auto";
    next_btn.style.opacity = step === steps.length-1 ? "0.3" : "1";
    next_btn.style.pointerEvents = step === steps.length-1 ? "none" : "auto";
    // Update dots
    for (let i = 0; i < dots_el.children.length; i++) {
      dots_el.children[i].style.background = i === step ? "#63b3ed" : "#4a5568";
      dots_el.children[i].style.transform = i === step ? "scale(1.4)" : "scale(1)";
    }
  }

  window[id + "_nav"] = function(dir) {
    step = Math.max(0, Math.min(steps.length-1, step + dir));
    render();
  };

  // Keyboard support
  document.addEventListener("keydown", function(e) {
    const rect = document.getElementById(id).getBoundingClientRect();
    const inView = rect.top < window.innerHeight && rect.bottom > 0;
    if (!inView) return;
    if (e.key === "ArrowRight" || e.key === " ") { e.preventDefault(); window[id + "_nav"](1); }
    if (e.key === "ArrowLeft") { e.preventDefault(); window[id + "_nav"](-1); }
  });

  render();
})();
</script>
"""))
````

### --- Arithmetic ---


````python
base_fare    = 18.50
passengers   = 43
stops        = 11
capacity     = 60

extra_stops  = stops - 5               # stops beyond the first 5
extra_charge = extra_stops * 2.50      # ₹2.50 per extra stop
total_fare   = base_fare + extra_charge
load_pct     = (passengers / capacity) * 100   # bus load percentage
stop_blocks  = stops // 3              # every 3 stops = 1 pricing block
leftover     = stops % 3               # stops that don't fill a full block

print("=== Arithmetic Results ===")
print(f"Extra stops  : {extra_stops}")
print(f"Extra charge : ₹{extra_charge}")
print(f"Total fare   : ₹{total_fare}")
print(f"Bus load     : {load_pct:.1f}%")
print(f"Pricing blocks: {stop_blocks}, leftover stops: {leftover}")
````

### --- Comparison ---


````python
is_crowded       = passengers > 40         # True
is_free          = total_fare == 0         # False
needs_extra_bus  = load_pct >= 90          # True/False depending on load

print("\n=== Comparison Results ===")
print(f"Is crowded?       {is_crowded}")
print(f"Is free ride?     {is_free}")
print(f"Needs extra bus?  {needs_extra_bus}")
````

### --- Logical ---


````python
is_peak          = True
is_student       = False
is_senior        = True

apply_surge      = is_peak and is_crowded          # surge: peak AND crowded
apply_concession = is_student or is_senior         # concession: either
full_fare        = not (is_student or is_senior)   # full fare if neither

print("\n=== Logical Results ===")
print(f"Apply surge?      {apply_surge}")
print(f"Apply concession? {apply_concession}")
print(f"Full fare?        {full_fare}")
````

### --- Assignment shortcuts ---


````python
daily_revenue = 0
daily_revenue += 795.50    # morning trips
daily_revenue += 1240.00   # afternoon trips
daily_revenue -= 50.00     # refunds
daily_revenue *= 1.05      # apply 5% GST
print(f"\nDaily revenue (incl. GST): ₹{daily_revenue:.2f}")
````

Build the CitySmart fare calculation engine using operators. No functions yet — just variables and operators.

### EXERCISE 1.2 — CitySmart Fare Calculation Engine (Mandatory)
#### ============================================================

````python
# Given trip details:
route_id        = 5
passengers      = 67
stops_travelled = 14
bus_capacity    = 80
base_fare       = 20.00
is_ac           = True
is_peak         = True
student_count   = 12    # of the 67 passengers, 12 are students
senior_count    = 8     # 8 are senior citizens
````

### --- TASK 1: Basic fare components ---


````python
# Extra stops cost ₹3 each (beyond the first 5 stops)
extra_stops     += 3
extra_fare      = extra_stops * 3.00
full_adult_fare = base_fare + extra_fare   # base_fare + extra_fare
````

### --- TASK 2: Surcharges ---


````python
# AC surcharge: ₹8 flat (only if AC bus)
# Peak surcharge: 15% on the full_adult_fare (only if peak hour)
ac_surcharge    = is_ac * 8   # use a logical expression — no if/else yet!
                        # Hint: True * 8 == 8, False * 8 == 0
peak_surcharge  = is_peak * full_adult_fare *0.15  # 15% of full_adult_fare, only if peak
total_adult_fare = full_adult_fare + ac_surcharge + peak_surcharge  # full_adult_fare + ac_surcharge + peak_surcharge
````

### ---TASK 3: Concession fares ---


````python

# Students pay 50% of total_adult_fare
# Seniors pay 40% of total_adult_fare
regular_passengers = passengers - student_count - senior_count
student_fare    = 0.50 * total_adult_fare   # 50% of total_adult_fare
senior_fare     =  0.40 * total_adult_fare   # 40% of total_adult_fare
````

### --- TASK 4: Total trip revenue ---





````python
revenue_regular  =  regular_passengers * total_adult_fare  # regular_passengers × total_adult_fare
revenue_students = student_count * student_fare # student_count × student_fare
revenue_seniors  = senior_count * senior_fare  # senior_count × senior_fare
total_revenue    = revenue_regular + revenue_seniors + revenue_students  # sum of all three
````

### --- TASK 5: Load analysis ---




````python
load_percentage = (passengers / capacity) * 100   # passengers as % of capacity
is_overcrowded  = load_percentage  # True if load_percentage > 100
is_surge_zone   = is_peak and load_percentage > 80   # True if is_peak AND load_percentage > 80
````

````python
# Print your results:
print(f"Route {route_id} | {passengers} passengers | {stops_travelled} stops")
print(f"Adult fare : ₹{total_adult_fare:.2f}")
print(f"Student fare: ₹{student_fare:.2f}")
print(f"Senior fare : ₹{senior_fare:.2f}")
print(f"Total revenue: ₹{total_revenue:.2f}")
print(f"Load: {load_percentage:.1f}% | Overcrowded: {is_overcrowded} | Surge zone: {is_surge_zone}")
````



### ✂️ Slicing — extracting substrings

 Slicing is the process of extracting a specific portion (substring) from a string by specifying a range of indices. It is a built-in shortcut for text manipulation that allows you to access parts of a sequence without manual looping.


````python
#@title ✂️ String Slicing (click ▶ to re-render if needed)
from IPython.display import HTML, display
display(HTML("""
<div id="slice_widget" style="
  max-width:620px; margin:16px auto; font-family:'Segoe UI',system-ui,sans-serif;
  background:linear-gradient(135deg,#1a1a2e 0%,#16213e 50%,#0f3460 100%);
  border-radius:14px; border:1px solid rgba(99,179,237,0.15);
  overflow:hidden; user-select:none;
">
  <!-- Title bar -->
  <div style="padding:14px 20px 6px; display:flex; justify-content:space-between; align-items:center;">
    <span style="color:#90cdf4; font-size:13px; letter-spacing:1.5px; text-transform:uppercase; font-weight:600;">✂️ String Slicing</span>
    <span id="slice_widget_counter" style="color:#4a5568; font-size:12px; font-family:monospace;"></span>
  </div>

  <!-- Visual area -->
  <div id="slice_widget_stage" style="padding:0 10px; min-height:340px; display:flex; align-items:center; justify-content:center;">
  </div>

  <!-- Controls -->
  <div style="padding:10px 20px 14px; display:flex; justify-content:center; align-items:center; gap:12px;">
    <button id="slice_widget_prev" onclick="slice_widget_nav(-1)" style="
      background:rgba(99,179,237,0.1); border:1px solid rgba(99,179,237,0.3); color:#90cdf4;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.25)'" onmouseout="this.style.background='rgba(99,179,237,0.1)'">
      ◀ Prev
    </button>

    <div id="slice_widget_dots" style="display:flex; gap:6px; align-items:center;"></div>

    <button id="slice_widget_next" onclick="slice_widget_nav(1)" style="
      background:rgba(99,179,237,0.2); border:1px solid rgba(99,179,237,0.4); color:#fff;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.4)'" onmouseout="this.style.background='rgba(99,179,237,0.2)'">
      Next ▶
    </button>
  </div>
</div>

<script>
(function() {
  const W = 600, H = 340;
  const id = "slice_widget";
  let step = 0;
  const steps = [`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="28" fill="#b794f4" font-size="22" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><text x="300" y="56" fill="#718096" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Extract substrings using [start : stop : step]</text><rect x="120" y="85" width="360" height="50" rx="8" fill="rgba(183,148,244,0.08)" stroke="rgba(183,148,244,0.3)" stroke-width="2" opacity="1"/><text x="155" y="112" fill="#a0aec0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">string</text><text x="228" y="112" fill="#fff" font-size="16" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">[</text><text x="240" y="112" fill="#63b3ed" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">start</text><text x="290" y="112" fill="#fff" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">:</text><text x="302" y="112" fill="#9ae6b4" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">stop</text><text x="342" y="112" fill="#fff" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">:</text><text x="354" y="112" fill="#fbd38d" font-size="15" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">step</text><text x="400" y="112" fill="#fff" font-size="16" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">]</text><text x="130" y="160" fill="#63b3ed" font-size="13" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">start</text><text x="190" y="160" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">— where to begin (included)</text><text x="130" y="182" fill="#9ae6b4" font-size="13" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">stop</text><text x="190" y="182" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">— where to end (EXCLUDED!)</text><text x="130" y="204" fill="#fbd38d" font-size="13" text-anchor="start" font-weight="normal" font-family="monospace" dominant-baseline="middle">step</text><text x="190" y="204" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">— skip count (default: 1)</text><text x="300" y="240" fill="#fc8181" font-size="12" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">The stop index is NOT included — this is a common gotcha!</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[0:3]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "12A"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Start to 3 (exclusive) — the route code</text><text x="63" y="84" fill="#63b3ed" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">start=0</text><line x1="58" y1="89" x2="58" y2="134" stroke="#63b3ed" stroke-width="2"/><text x="162" y="84" fill="#fc8181" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">stop=3</text><line x1="190" y1="89" x2="190" y2="134" stroke="rgba(252,129,129,0.5)" stroke-width="2"/><circle cx="262" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[4:]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "Express"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Index 4 to end — the route type</text><text x="239" y="84" fill="#63b3ed" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">start=4</text><line x1="234" y1="89" x2="234" y2="134" stroke="#63b3ed" stroke-width="2"/><circle cx="262" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[:3]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "12A"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Same as [0:3] — omitting start means 0</text><text x="63" y="84" fill="#63b3ed" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">start=0</text><line x1="58" y1="89" x2="58" y2="134" stroke="#63b3ed" stroke-width="2"/><text x="162" y="84" fill="#fc8181" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">stop=3</text><line x1="190" y1="89" x2="190" y2="134" stroke="rgba(252,129,129,0.5)" stroke-width="2"/><circle cx="262" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[-7:]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "Express"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Last 7 characters using negative index</text><text x="239" y="84" fill="#63b3ed" font-size="10" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">start=4</text><line x1="234" y1="89" x2="234" y2="134" stroke="#63b3ed" stroke-width="2"/><circle cx="262" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(74,85,104,0.08)" stroke="#4a5568" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#4a5568" font-size="18" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[0:7:2]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "1AEx"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Every 2nd char from 0 to 7 (step=2)</text><circle cx="262" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 340" width="600" height="340" style="font-family:monospace;"><text x="300" y="24" fill="#b794f4" font-size="20" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">String Slicing</text><rect x="58" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="79" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">1</text><rect x="102" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="123" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">2</text><rect x="146" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="167" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">A</text><rect x="190" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="211" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle"> </text><rect x="234" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="255" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">E</text><rect x="278" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="299" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">x</text><rect x="322" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="343" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">p</text><rect x="366" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="387" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">r</text><rect x="410" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="431" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">e</text><rect x="454" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="475" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><rect x="498" y="90" width="41" height="44" rx="4" fill="rgba(154,230,180,0.25)" stroke="#9ae6b4" stroke-width="2" opacity="1"/><text x="519" y="114" fill="#fff" font-size="18" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">s</text><text x="79" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">0</text><text x="123" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">1</text><text x="167" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">2</text><text x="211" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">3</text><text x="255" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">4</text><text x="299" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">5</text><text x="343" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">6</text><text x="387" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">7</text><text x="431" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">8</text><text x="475" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">9</text><text x="519" y="148" fill="#4a5568" font-size="11" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">10</text><text x="80" y="190" fill="#e2e8f0" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">route_name[::-1]</text><text x="380" y="190" fill="#9ae6b4" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">→ "sserpxE A21"</text><text x="300" y="225" fill="#a0aec0" font-size="13" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Reversed! step=-1 reads backwards</text><circle cx="262" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="278" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="294" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="310" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="326" cy="280" r="3" fill="#4a5568" stroke="none" stroke-width="0"/><circle cx="342" cy="280" r="4" fill="#9ae6b4" stroke="none" stroke-width="0"/></svg>`];

  const stage = document.getElementById(id + "_stage");
  const counter = document.getElementById(id + "_counter");
  const dots_el = document.getElementById(id + "_dots");
  const prev_btn = document.getElementById(id + "_prev");
  const next_btn = document.getElementById(id + "_next");

  // Build dots
  for (let i = 0; i < steps.length; i++) {
    const d = document.createElement("div");
    d.style.cssText = "width:8px;height:8px;border-radius:50%;background:#4a5568;transition:all 0.2s;cursor:pointer;";
    d.onclick = function() { step = i; render(); };
    dots_el.appendChild(d);
  }

  function render() {
    stage.innerHTML = steps[step];
    counter.textContent = "Step " + (step+1) + " of " + steps.length;
    prev_btn.style.opacity = step === 0 ? "0.3" : "1";
    prev_btn.style.pointerEvents = step === 0 ? "none" : "auto";
    next_btn.style.opacity = step === steps.length-1 ? "0.3" : "1";
    next_btn.style.pointerEvents = step === steps.length-1 ? "none" : "auto";
    // Update dots
    for (let i = 0; i < dots_el.children.length; i++) {
      dots_el.children[i].style.background = i === step ? "#63b3ed" : "#4a5568";
      dots_el.children[i].style.transform = i === step ? "scale(1.4)" : "scale(1)";
    }
  }

  window[id + "_nav"] = function(dir) {
    step = Math.max(0, Math.min(steps.length-1, step + dir));
    render();
  };

  // Keyboard support
  document.addEventListener("keydown", function(e) {
    const rect = document.getElementById(id).getBoundingClientRect();
    const inView = rect.top < window.innerHeight && rect.bottom > 0;
    if (!inView) return;
    if (e.key === "ArrowRight" || e.key === " ") { e.preventDefault(); window[id + "_nav"](1); }
    if (e.key === "ArrowLeft") { e.preventDefault(); window[id + "_nav"](-1); }
  });

  render();
})();
</script>
"""))
````


### 💬 f-strings — the modern way to format strings

```python
  # Old way (ugly)
  print("Route " + str(route_id) + " has " + str(stops) + " stops")

  # f-string way (clean, powerful)
  print(f"Route {route_id} has {stops} stops")
  print(f"Fare: ₹{fare:.2f}")         # .2f = 2 decimal places
  print(f"Load: {load_pct:.1%}")       # .1% = percentage format
  print(f"Route: {route_id:>6}")       # >6 = right-align in 6 chars
  print(f"Route: {route_id:<6}")       # <6 = left-align in 6 chars
  print(f"Route: {route_id:0>4}")      # pad with zeros: 0007
```

### 🛠 The 10 string methods every CitySmart developer uses

```
  Method            What it does                      Example
  ────────────────  ────────────────────────────────  ─────────────────────────────
  .upper()          ALL CAPS                          "12a express".upper() → "12A EXPRESS"
  .lower()          all lowercase                     "12A EXPRESS".lower() → "12a express"
  .strip()          remove leading/trailing spaces    " Central ".strip()  → "Central"
  .split(sep)       split into a list                 "A,B,C".split(",")   → ['A','B','C']
  .join(list)       join list into string             ",".join(['A','B'])   → "A,B"
  .replace(a,b)     replace all occurrences           "bus stop".replace("stop","stand")
  .find(sub)        index of first match (-1 if none) "Express".find("press") → 2
  .startswith(s)    does it start with s?             "12A".startswith("12") → True
  .endswith(s)      does it end with s?               "Express".endswith("ss") → True
  .count(sub)       how many times does sub appear?   "abcabc".count("bc")   → 2
```

````python
# Format it like this:
#
# ┌─────────────────────────────────────────┐
#  Route 007 · Express · North
#  Stops (5): Central Bus Stand → Mg Road
#             → Silk Board → Electronic City
#             → Bommanahalli
# └─────────────────────────────────────────┘
#
# (The "→" join between stops should be done with .join())

# Write your formatted print statements here:


def print_route_card(num, rtype, direction, stops, variant=""):
    print("┌" + "─" * 43 + "┐")
    print(f"  Route {num:03d}{variant} · {rtype} · {direction}")
    head = f"  Stops ({len(stops)}): "
    print(head + " → ".join(stops[:2]))
    for s in stops[2:]:
        print(" " * len(head) + "→ " + s)
    print("└" + "─" * 43 + "┘")

print_route_card(route_num_1, route_type_1, direction_1, stops_list_1)
print_route_card(route_num_2, route_type_2, direction_2, stops_list_2, variant=route_var_2)
print_route_card(42, "Harbour Line", "—", stops_list_3)
````

### Python Conditional Statements: `if`, `elif`, and `else`

In Python, `if`, `elif`, and `else` are conditional statements used for **decision-making**, allowing a program to execute specific blocks of code based on whether certain conditions are `True` or `False`.

---

## Key Components

### `if` Statement
The **starting point** of any decision-making block. It evaluates a condition; if it is `True`, the indented code block directly below it runs.



---

### `elif` Statement
Short for **"else if"** — allows you to check multiple alternative conditions if the initial `if` condition was `False`. You can use multiple `elif` blocks to handle various scenarios.



---

### `else` Statement
Acts as a **"catch-all"** or fallback. It executes only if **all** preceding `if` and `elif` conditions evaluate to `False`.


---



> **Tip:** Only **one** block executes per `if`/`elif`/`else` chain — Python stops checking as soon as it finds the first `True` condition.

````python
#@title 🚦 Control Flow (click ▶ to re-render if needed)
from IPython.display import HTML, display
display(HTML("""
<div id="cf_widget" style="
  max-width:620px; margin:16px auto; font-family:'Segoe UI',system-ui,sans-serif;
  background:linear-gradient(135deg,#1a1a2e 0%,#16213e 50%,#0f3460 100%);
  border-radius:14px; border:1px solid rgba(99,179,237,0.15);
  overflow:hidden; user-select:none;
">
  <!-- Title bar -->
  <div style="padding:14px 20px 6px; display:flex; justify-content:space-between; align-items:center;">
    <span style="color:#90cdf4; font-size:13px; letter-spacing:1.5px; text-transform:uppercase; font-weight:600;">🚦 Control Flow: if/elif/else</span>
    <span id="cf_widget_counter" style="color:#4a5568; font-size:12px; font-family:monospace;"></span>
  </div>

  <!-- Visual area -->
  <div id="cf_widget_stage" style="padding:0 10px; min-height:360px; display:flex; align-items:center; justify-content:center;">
  </div>

  <!-- Controls -->
  <div style="padding:10px 20px 14px; display:flex; justify-content:center; align-items:center; gap:12px;">
    <button id="cf_widget_prev" onclick="cf_widget_nav(-1)" style="
      background:rgba(99,179,237,0.1); border:1px solid rgba(99,179,237,0.3); color:#90cdf4;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.25)'" onmouseout="this.style.background='rgba(99,179,237,0.1)'">
      ◀ Prev
    </button>

    <div id="cf_widget_dots" style="display:flex; gap:6px; align-items:center;"></div>

    <button id="cf_widget_next" onclick="cf_widget_nav(1)" style="
      background:rgba(99,179,237,0.2); border:1px solid rgba(99,179,237,0.4); color:#fff;
      padding:8px 20px; border-radius:8px; cursor:pointer; font-size:14px; font-weight:600;
      transition:all 0.15s;
    " onmouseover="this.style.background='rgba(99,179,237,0.4)'" onmouseout="this.style.background='rgba(99,179,237,0.2)'">
      Next ▶
    </button>
  </div>
</div>

<script>
(function() {
  const W = 600, H = 360;
  const id = "cf_widget";
  let step = 0;
  const steps = [`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 360" width="600" height="360" style="font-family:monospace;"><text x="300" y="12" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Decision Tree</text><line x1="300" y1="59" x2="300" y2="91" stroke="#4a5568" stroke-width="1"/><line x1="300" y1="119" x2="120" y2="161" stroke="#4a5568" stroke-width="1"/><text x="198" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="300" y1="119" x2="380" y2="161" stroke="#4a5568" stroke-width="1"/><text x="328" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="380" y1="189" x2="280" y2="231" stroke="#4a5568" stroke-width="1"/><text x="318" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="380" y1="189" x2="480" y2="231" stroke="#4a5568" stroke-width="1"/><text x="418" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="480" y1="259" x2="410" y2="296" stroke="#4a5568" stroke-width="1"/><text x="433" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="480" y1="259" x2="550" y2="296" stroke="#4a5568" stroke-width="1"/><text x="503" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><rect x="226" y="31" width="148" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="300" y="46" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Passenger boards</text><rect x="246" y="91" width="108" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="300" y="106" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_student?</text><rect x="78" y="161" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="120" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">50% fare</text><rect x="330" y="161" width="100" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="380" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_senior?</text><rect x="238" y="231" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="280" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">40% fare</text><rect x="434" y="231" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="480" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_staff?</text><rect x="364" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="410" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Free ride</text><rect x="504" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="550" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Full fare</text><text x="300" y="340" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">A passenger boards the bus — which fare do they pay?</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 360" width="600" height="360" style="font-family:monospace;"><text x="300" y="12" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Decision Tree</text><line x1="300" y1="59" x2="300" y2="91" stroke="#4a5568" stroke-width="1"/><line x1="300" y1="119" x2="120" y2="161" stroke="#4a5568" stroke-width="1"/><text x="198" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="300" y1="119" x2="380" y2="161" stroke="#4a5568" stroke-width="1"/><text x="328" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="380" y1="189" x2="280" y2="231" stroke="#4a5568" stroke-width="1"/><text x="318" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="380" y1="189" x2="480" y2="231" stroke="#4a5568" stroke-width="1"/><text x="418" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="480" y1="259" x2="410" y2="296" stroke="#4a5568" stroke-width="1"/><text x="433" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="480" y1="259" x2="550" y2="296" stroke="#4a5568" stroke-width="1"/><text x="503" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><rect x="226" y="31" width="148" height="28" rx="8" fill="rgba(99,179,237,0.4)" stroke="rgb(99,179,237)" stroke-width="3" opacity="1"/><text x="300" y="46" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">Passenger boards</text><circle cx="300" cy="45" r="20" fill="rgba(118,228,247,0.15)" stroke="none" stroke-width="0"/><circle cx="300" cy="45" r="14" fill="rgba(118,228,247,0.25)" stroke="none" stroke-width="0"/><rect x="246" y="91" width="108" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="300" y="106" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_student?</text><rect x="78" y="161" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="120" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">50% fare</text><rect x="330" y="161" width="100" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="380" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_senior?</text><rect x="238" y="231" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="280" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">40% fare</text><rect x="434" y="231" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="480" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_staff?</text><rect x="364" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="410" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Free ride</text><rect x="504" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="550" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Full fare</text><text x="300" y="340" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Passenger boards → check conditions from top to bottom</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 360" width="600" height="360" style="font-family:monospace;"><text x="300" y="12" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Decision Tree</text><line x1="300" y1="59" x2="300" y2="91" stroke="#76e4f7" stroke-width="3"/><line x1="300" y1="119" x2="120" y2="161" stroke="#4a5568" stroke-width="1"/><text x="198" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="300" y1="119" x2="380" y2="161" stroke="#4a5568" stroke-width="1"/><text x="328" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="380" y1="189" x2="280" y2="231" stroke="#4a5568" stroke-width="1"/><text x="318" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="380" y1="189" x2="480" y2="231" stroke="#4a5568" stroke-width="1"/><text x="418" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="480" y1="259" x2="410" y2="296" stroke="#4a5568" stroke-width="1"/><text x="433" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="480" y1="259" x2="550" y2="296" stroke="#4a5568" stroke-width="1"/><text x="503" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><rect x="226" y="31" width="148" height="28" rx="8" fill="rgba(99,179,237,0.2)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="300" y="46" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">Passenger boards</text><rect x="246" y="91" width="108" height="28" rx="8" fill="rgba(183,148,244,0.4)" stroke="rgb(183,148,244)" stroke-width="3" opacity="1"/><text x="300" y="106" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">is_student?</text><circle cx="300" cy="105" r="20" fill="rgba(118,228,247,0.15)" stroke="none" stroke-width="0"/><circle cx="300" cy="105" r="14" fill="rgba(118,228,247,0.25)" stroke="none" stroke-width="0"/><rect x="78" y="161" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="120" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">50% fare</text><rect x="330" y="161" width="100" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="380" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_senior?</text><rect x="238" y="231" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="280" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">40% fare</text><rect x="434" y="231" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="480" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_staff?</text><rect x="364" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="410" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Free ride</text><rect x="504" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="550" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Full fare</text><text x="300" y="340" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">if is_student: → False! Move to elif...</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 360" width="600" height="360" style="font-family:monospace;"><text x="300" y="12" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Decision Tree</text><line x1="300" y1="59" x2="300" y2="91" stroke="#76e4f7" stroke-width="3"/><line x1="300" y1="119" x2="120" y2="161" stroke="#4a5568" stroke-width="1"/><text x="198" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="300" y1="119" x2="380" y2="161" stroke="#76e4f7" stroke-width="3"/><text x="328" y="135" fill="#fc8181" font-size="11" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="380" y1="189" x2="280" y2="231" stroke="#4a5568" stroke-width="1"/><text x="318" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="380" y1="189" x2="480" y2="231" stroke="#4a5568" stroke-width="1"/><text x="418" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="480" y1="259" x2="410" y2="296" stroke="#4a5568" stroke-width="1"/><text x="433" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="480" y1="259" x2="550" y2="296" stroke="#4a5568" stroke-width="1"/><text x="503" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><rect x="226" y="31" width="148" height="28" rx="8" fill="rgba(99,179,237,0.2)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="300" y="46" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">Passenger boards</text><rect x="246" y="91" width="108" height="28" rx="8" fill="rgba(183,148,244,0.2)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="300" y="106" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">is_student?</text><rect x="78" y="161" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="120" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">50% fare</text><rect x="330" y="161" width="100" height="28" rx="8" fill="rgba(183,148,244,0.4)" stroke="rgb(183,148,244)" stroke-width="3" opacity="1"/><text x="380" y="176" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">is_senior?</text><circle cx="380" cy="175" r="20" fill="rgba(118,228,247,0.15)" stroke="none" stroke-width="0"/><circle cx="380" cy="175" r="14" fill="rgba(118,228,247,0.25)" stroke="none" stroke-width="0"/><rect x="238" y="231" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="280" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">40% fare</text><rect x="434" y="231" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="480" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_staff?</text><rect x="364" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="410" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Free ride</text><rect x="504" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="550" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Full fare</text><text x="300" y="340" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">elif is_senior: → True! Execute this branch.</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 360" width="600" height="360" style="font-family:monospace;"><text x="300" y="12" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Decision Tree</text><line x1="300" y1="59" x2="300" y2="91" stroke="#76e4f7" stroke-width="3"/><line x1="300" y1="119" x2="120" y2="161" stroke="#4a5568" stroke-width="1"/><text x="198" y="135" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="300" y1="119" x2="380" y2="161" stroke="#76e4f7" stroke-width="3"/><text x="328" y="135" fill="#fc8181" font-size="11" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="380" y1="189" x2="280" y2="231" stroke="#76e4f7" stroke-width="3"/><text x="318" y="205" fill="#9ae6b4" font-size="11" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="380" y1="189" x2="480" y2="231" stroke="#4a5568" stroke-width="1"/><text x="418" y="205" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><line x1="480" y1="259" x2="410" y2="296" stroke="#4a5568" stroke-width="1"/><text x="433" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">Yes</text><line x1="480" y1="259" x2="550" y2="296" stroke="#4a5568" stroke-width="1"/><text x="503" y="272" fill="#4a5568" font-size="11" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">No</text><rect x="226" y="31" width="148" height="28" rx="8" fill="rgba(99,179,237,0.2)" stroke="rgb(99,179,237)" stroke-width="2" opacity="1"/><text x="300" y="46" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">Passenger boards</text><rect x="246" y="91" width="108" height="28" rx="8" fill="rgba(183,148,244,0.2)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="300" y="106" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">is_student?</text><rect x="78" y="161" width="84" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="120" y="176" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">50% fare</text><rect x="330" y="161" width="100" height="28" rx="8" fill="rgba(183,148,244,0.2)" stroke="rgb(183,148,244)" stroke-width="2" opacity="1"/><text x="380" y="176" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">is_senior?</text><rect x="238" y="231" width="84" height="28" rx="8" fill="rgba(154,230,180,0.4)" stroke="rgb(154,230,180)" stroke-width="3" opacity="1"/><text x="280" y="246" fill="#fff" font-size="12" text-anchor="middle" font-weight="bold" font-family="monospace" dominant-baseline="middle">40% fare</text><circle cx="280" cy="245" r="20" fill="rgba(118,228,247,0.15)" stroke="none" stroke-width="0"/><circle cx="280" cy="245" r="14" fill="rgba(118,228,247,0.25)" stroke="none" stroke-width="0"/><rect x="434" y="231" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="480" y="246" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">is_staff?</text><rect x="364" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="410" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Free ride</text><rect x="504" y="296" width="92" height="28" rx="8" fill="rgba(74,85,104,0.1)" stroke="#4a5568" stroke-width="1" opacity="1"/><text x="550" y="311" fill="#718096" font-size="12" text-anchor="middle" font-weight="normal" font-family="monospace" dominant-baseline="middle">Full fare</text><text x="300" y="340" fill="#a0aec0" font-size="12" text-anchor="middle" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">✅ fare = 40% — remaining branches are SKIPPED entirely!</text></svg>`,`<svg xmlns="http://www.w3.org/2000/svg" viewBox="0 0 600 320" width="600" height="320" style="font-family:monospace;"><text x="300" y="20" fill="#63b3ed" font-size="18" text-anchor="middle" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">if / elif / else — Key Rules</text><rect x="50" y="65" width="500" height="48" rx="8" fill="rgba(99,179,237,0.06)" stroke="rgba(99,179,237,0.2)" stroke-width="1" opacity="1"/><text x="75" y="83" fill="rgb(99,179,237)" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">1.</text><text x="100" y="83" fill="#e2e8f0" font-size="15" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Only ONE branch executes</text><text x="100" y="103" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">the first True condition wins</text><rect x="50" y="125" width="500" height="48" rx="8" fill="rgba(154,230,180,0.06)" stroke="rgba(154,230,180,0.2)" stroke-width="1" opacity="1"/><text x="75" y="143" fill="rgb(154,230,180)" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">2.</text><text x="100" y="143" fill="#e2e8f0" font-size="15" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">elif checks are skipped</text><text x="100" y="163" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">if any earlier condition was True</text><rect x="50" y="185" width="500" height="48" rx="8" fill="rgba(251,211,141,0.06)" stroke="rgba(251,211,141,0.2)" stroke-width="1" opacity="1"/><text x="75" y="203" fill="rgb(251,211,141)" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">3.</text><text x="100" y="203" fill="#e2e8f0" font-size="15" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">else is the fallback</text><text x="100" y="223" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">runs only if NOTHING matched</text><rect x="50" y="245" width="500" height="48" rx="8" fill="rgba(252,129,129,0.06)" stroke="rgba(252,129,129,0.2)" stroke-width="1" opacity="1"/><text x="75" y="263" fill="rgb(252,129,129)" font-size="16" text-anchor="start" font-weight="bold" font-family="monospace" dominant-baseline="middle">4.</text><text x="100" y="263" fill="#e2e8f0" font-size="15" text-anchor="start" font-weight="bold" font-family="sans-serif" dominant-baseline="middle">Indentation matters!</text><text x="100" y="283" fill="#a0aec0" font-size="12" text-anchor="start" font-weight="normal" font-family="sans-serif" dominant-baseline="middle">4 spaces define what's inside the block</text></svg>`];

  const stage = document.getElementById(id + "_stage");
  const counter = document.getElementById(id + "_counter");
  const dots_el = document.getElementById(id + "_dots");
  const prev_btn = document.getElementById(id + "_prev");
  const next_btn = document.getElementById(id + "_next");

  // Build dots
  for (let i = 0; i < steps.length; i++) {
    const d = document.createElement("div");
    d.style.cssText = "width:8px;height:8px;border-radius:50%;background:#4a5568;transition:all 0.2s;cursor:pointer;";
    d.onclick = function() { step = i; render(); };
    dots_el.appendChild(d);
  }

  function render() {
    stage.innerHTML = steps[step];
    counter.textContent = "Step " + (step+1) + " of " + steps.length;
    prev_btn.style.opacity = step === 0 ? "0.3" : "1";
    prev_btn.style.pointerEvents = step === 0 ? "none" : "auto";
    next_btn.style.opacity = step === steps.length-1 ? "0.3" : "1";
    next_btn.style.pointerEvents = step === steps.length-1 ? "none" : "auto";
    // Update dots
    for (let i = 0; i < dots_el.children.length; i++) {
      dots_el.children[i].style.background = i === step ? "#63b3ed" : "#4a5568";
      dots_el.children[i].style.transform = i === step ? "scale(1.4)" : "scale(1)";
    }
  }

  window[id + "_nav"] = function(dir) {
    step = Math.max(0, Math.min(steps.length-1, step + dir));
    render();
  };

  // Keyboard support
  document.addEventListener("keydown", function(e) {
    const rect = document.getElementById(id).getBoundingClientRect();
    const inView = rect.top < window.innerHeight && rect.bottom > 0;
    if (!inView) return;
    if (e.key === "ArrowRight" || e.key === " ") { e.preventDefault(); window[id + "_nav"](1); }
    if (e.key === "ArrowLeft") { e.preventDefault(); window[id + "_nav"](-1); }
  });

  render();
})();
</script>
"""))
````



### 🔁 for loops — iterate over a known sequence

```
  Use for when you know WHAT you're iterating over.

  for stop in route_stops:          ← iterate over a list
  for i in range(10):               ← iterate 0 to 9
  for i, stop in enumerate(stops):  ← get index AND value
  for route, fare in fare_table.items(): ← dict key & value
```

### ♾ while loops — iterate until a condition changes

```
  Use while when you don't know HOW MANY times to iterate.

  waiting = True
  while waiting:                    ← keep checking
      bus = check_next_bus()
      if bus.route == "12A":
          waiting = False           ← condition met, exit

  ⚠ DANGER: Infinite loops — always make sure the condition
    can become False, or use break to exit!
```

### ⏩ break and continue

```
  break     → exit the loop immediately, skip the rest
  continue  → skip the rest of THIS iteration, go to next

  for trip in today_trips:
      if trip.passengers == 0:
          continue        ← skip empty trips (no revenue to count)
      if trip.is_cancelled:
          break           ← stop processing after first cancellation
      process(trip)
```

````python
# The CitySmart route data (we'll use proper data structures next lesson)
routes = {
    "7":  {"name": "Harbour Express",   "stops": ["Central", "Fort", "Harbour", "Navy Nagar"],          "base_fare": 20.0, "is_ac": True},
    "12": {"name": "Airport Shuttle",   "stops": ["Majestic", "MG Road", "Airport T1", "Airport T2"],  "base_fare": 35.0, "is_ac": True},
    "31": {"name": "Local City Loop",   "stops": ["KR Market", "Shivajinagar", "Ulsoor", "Richmond","KR Market"], "base_fare": 12.0, "is_ac": False},
    "42": {"name": "North Connector",   "stops": ["Yeshwanthpur", "Peenya", "Tumkur Road", "Nelamangala"], "base_fare": 18.0, "is_ac": False},
}
````

### --- if/elif/else: Fare category decision ---


````python
print("=== Fare Category Logic ===")
# Try changing these flags and re-running to see different outcomes
is_student  = False
is_senior   = True
is_staff    = False
base_fare   = 35.0

if is_student:
    discount = 0.50
    category = "Student (50% off)"
elif is_senior:
    discount = 0.40
    category = "Senior Citizen (40% off)"
elif is_staff:
    discount = 1.00
    category = "Staff (free)"
else:
    discount = 0.00
    category = "Full fare"

final_fare = base_fare * (1 - discount)
print(f"Category : {category}")
print(f"Fare     : ₹{final_fare:.2f}")
````

### --- for loop: Print all stops on a route ---


````python
print("\n=== Route 12 Stop Schedule ===")
route_12_stops = routes["12"]["stops"]

for i, stop in enumerate(route_12_stops):
    if i == 0:
        label = "[ORIGIN]"
    elif i == len(route_12_stops) - 1:
        label = "[TERMINUS]"
    else:
        label = f"[Stop {i}]"
    print(f"  {label:<12} {stop}")
````

### --- for with break/continue: find the first AC route ---


````python

print("\n=== Finding First AC Route ===")
for route_id, info in routes.items():
    if not info["is_ac"]:
        continue           # skip non-AC routes
    print(f"First AC route found: Route {route_id} — {info['name']}")
    break                  # stop after the first match
````

### --- while loop: simulate passengers boarding until full ---


````python
print("\n=== Boarding Simulation ===")
capacity   = 60
boarded    = 0
queue      = [8, 12, 15, 10, 9, 7, 6, 4, 3]  # passengers at each stop
stop_index = 0

while boarded < capacity and stop_index < len(queue):
    boarding_now  = queue[stop_index]
    space_left    = capacity - boarded
    actual_boards = min(boarding_now, space_left)  # can't exceed capacity
    boarded      += actual_boards

    print(f"  Stop {stop_index+1}: {actual_boards} board → Total: {boarded}/{capacity}")

    if boarded >= capacity:
        print("  ⚠ Bus is full — closing doors!")
        break
    stop_index += 1
````

### --- TASK 3: Find routes that serve a given stop ---


````python
# User wants to get to "MG Road" — which routes serve it?
search_stop = "MG Road"
found_routes = []

for route_id, info in routes.items():                # loop over all routes
    for stop in info["stops"]:            # loop over that route's stops
        if stop == search_stop:         # if this stop matches the search
            found_routes.append(route_id)         # add the route_id to found_routes
            break       # no need to keep checking this route's stops

if found_routes:
    print(f"\nRoutes serving '{search_stop}': {found_routes}")
else:
    print(f"\nNo routes found serving '{search_stop}'")
````

````python
def calculate_fare(passengers, stops, base_fare, is_student, is_ac, is_peak):
    """
    Calculate the total fare for a CitySmart trip.

    Rules:
    - Base fare covers the first 5 stops
    - Each additional stop: +₹2.50
    - Student discount: 50% off
    - Senior discount: 40% off  (not used here)
    - AC surcharge: +₹8 flat
    - Peak surcharge: +15% on fare (applied BEFORE concessions)
    - Total revenue = passengers * individual_fare
    """

    # Step 1: calculate fare for extra stops
    extra_stops  = stops - 5
    extra_charge = extra_stops * 2.50
    base_total   = base_fare + extra_charge

    # Step 2: apply peak surcharge
    if is_peak = True:                        # BUG ?
        base_total = base_total + base_total * 0.15

    # Step 3: apply AC surcharge
    if is_ac:
        base_total += 8

    # Step 4: apply student discount
    if is_student == True:
        individual_fare = base_total - base_total * 50  # BUG ?
    else:
        individual_fare = base_total

    # Step 5: calculate total revenue
    total_revenue = passengers * individual_fare

    return individual_fare, total_revenue


# Test case: 40 student passengers, 8 stops, ₹20 base, AC, peak hour
# Expected individual_fare: (20 + 7.50) * 1.15 + 8 = 27.50*1.15+8 = 31.625+8 = 39.625 * 0.50 = 19.81
# Expected total_revenue  : 40 * 19.81 = 792.50

fare, revenue = calculate_fare(
    passengers=40,
    stops=8,
    base_fare=20.0,
    is_student=True,
    is_ac=True,
    is_peak=True
)
print(f"Individual fare : ₹{fare:.2f}   (expected: ₹19.81)")
print(f"Total revenue   : ₹{revenue:.2f}  (expected: ₹792.50)")

````

## 🔑 Key concepts recap

| Concept | What it is | Where you used it |
|---|---|---|
| `int` / `float` / `str` / `bool` | The 4 basic types | Fare, stops, route name, AC flag |
| Dynamic typing | Python infers types | `fare = 18` then `fare = 18.5` |
| Type casting | `int()` `float()` `str()` `bool()` | Parsing CSV/input data |
| Arithmetic operators | `+ - * / // % **` | Fare calculation, load % |
| Comparison operators | `> < == != >= <=` | Crowded check, surge logic |
| Logical operators | `and or not` | Concession eligibility |
| Assignment shortcuts | `+= -= *= //=` | Accumulating revenue |
| String indexing | `s[0]`, `s[-1]` | Route code parsing |
| String slicing | `s[2:5]`, `s[::-1]` | Extract route number, type |
| f-strings | `f"Route {id}"` | All formatted output |
| String methods | `.strip()` `.split()` `.join()` etc | Data cleaning |
| `if/elif/else` | Decision branching | Fare category, menu routing |
| `for` loop | Iterate known sequences | Stop listing, route searching |
| `while` loop | Iterate until condition | Main menu loop, boarding sim |
| `break` / `continue` | Loop control | Skip empty, stop at first match |
